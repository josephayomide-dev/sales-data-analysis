```python
import pandas as pd
import matplotlib.pyplot as plt

sales_data = {
    "Date": [
        "2026-01-05", "2026-01-12", "2026-01-20",
        "2026-02-03", "2026-02-14", "2026-02-22",
        "2026-03-02", "2026-03-11", "2026-03-19",
        "2026-04-04", "2026-04-15", "2026-04-25"
    ],
    "Product": [
        "Rice", "Beans", "Cooking Oil",
        "Rice", "Cooking Oil", "Beans",
        "Rice", "Beans", "Cooking Oil",
        "Rice", "Cooking Oil", "Beans"
    ],
    "Quantity_Sold": [
        20, 15, 10,
        25, 12, 18,
        30, 20, 15,
        35, 18, 22
    ],
    "Unit_Price": [
        2000, 1500, 3500,
        2000, 3500, 1500,
        2000, 1500, 3500,
        2000, 3500, 1500
    ]
}

df = pd.DataFrame(sales_data)
df["Date"] = pd.to_datetime(df["Date"])
df["Revenue"] = df["Quantity_Sold"] * df["Unit_Price"]

df.head()
```

```python
print("Number of sales records:", len(df))

print("\nProducts sold:")
print(df["Product"].unique())

print("\nTotal revenue:")
print(df["Revenue"].sum())

print("\nTotal quantity sold:")
print(df["Quantity_Sold"].sum())
```

```python
product_sales = df.groupby("Product")["Quantity_Sold"].sum()

print("Total units sold by product:")
print(product_sales.sort_values(ascending=False))
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum()

print("Total revenue by product:")
print(product_revenue.sort_values(ascending=False))
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum()
product_revenue = product_revenue.sort_values(ascending=False)

plt.figure(figsize=(9, 5))

product_revenue.plot(kind="bar")

plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
df["Month"] = df["Date"].dt.month_name()

monthly_revenue = df.groupby("Month")["Revenue"].sum()

month_order = [
    "January", "February", "March", "April"
]

monthly_revenue = monthly_revenue.reindex(month_order)

print("Monthly revenue:")
print(monthly_revenue)
```

```python
plt.figure(figsize=(9, 5))

monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
average_sale = df.groupby("Product")["Revenue"].mean()

print("Average revenue per transaction by product:")
print(average_sale.round(2).sort_values(ascending=False))
```
